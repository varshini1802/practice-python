```python
#expence track

total = 0

while True:
    print("\n1. Add Expense")
    print("2. View Total")
    print("3. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        expense = float(input("Enter expense: "))

        file = open("expenses.txt", "a")
        file.write(str(expense) + "\n")
        file.close()

        total = total + expense

        print("Expense added")

    elif choice == 2:
        print("Total expense:", total)

    elif choice == 3:
        break

    else:
        print("Invalid choice")
```
