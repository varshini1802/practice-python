```python
#note app

while True:
    print("\n1. Add Note")
    print("2. View Notes")
    print("3. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        note = input("Enter your note: ")

        file = open("notes.txt", "a")
        file.write(note + "\n")
        file.close()

        print("Note saved")

    elif choice == 2:
        file = open("notes.txt", "r")

        print(file.read())

        file.close()

    elif choice == 3:
        break

    else:
        print("Invalid choice")
```
