```python
#Count the number of lines, words, and characters in a text file (Medium)

file=open("note.txt","r")
content=file.read()

lines= content.split("\n")
words=content.split()
characters=len(content)

print("Lines=",len(lines))
print("words=",len(words))
print("Charcters=",characters)

file.close()
```
