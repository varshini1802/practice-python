```python
#Build a shopping cart program using a list of dictionaries (item, price, quantity) (Medium)

cart=[
    {"item":"Apple", "price":20, "quantity":5},
    {"item":"Mango", "price":70, "quantity":2},
    {"item":"Watermelon", "price":50, "quantity":1}
]
total=0

for i in cart:
    total=total+i["price"]*i["quantity"]

print("Total bill:",total)
```
