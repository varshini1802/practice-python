```python
#Build a basic student database using a list of dictionaries with add, update, delete, search (Hard)

students = []

while True:
    print("\n1. Add")
    print("2. Update")
    print("3. Delete")
    print("4. Search")
    print("5. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        name = input("Enter name: ")
        marks = int(input("Enter marks: "))

        students.append({"name": name, "marks": marks})
        print("Student added")

    elif choice == 2:
        name = input("Enter name: ")

        for student in students:
            if student["name"] == name:
                student["marks"] = int(input("Enter new marks: "))
                print("Student updated")

    elif choice == 3:
        name = input("Enter name: ")

        for student in students:
            if student["name"] == name:
                students.remove(student)
                print("Student deleted")
                break

    elif choice == 4:
        name = input("Enter name: ")

        for student in students:
            if student["name"] == name:
                print(student)

    elif choice == 5:
        break

    else:
        print("Invalid choice")
```
