```python
#Build a menu-driven program that calls separate add/subtract/multiply/divide functions (Hard)

def add(a,b):
    return a+b
def sub(a,b):
    return a-b
def mul(a,b):
    return a*b
def div(a,b):
    return a/b

print("1. Add")
print("2. Subtract")
print("3. Multiply")
print("4. Divide")

a=int(input("Enter value of a"))
b=int(input("Enter value of b"))

choice=int(input("choose any (1,2,3,4)"))

if choice==1:
    print(add(a,b))
elif choice==2:
    print(sub(a,b))
elif choice==3:
    print(mul(a,b))
elif choice==4:
    print(div(a,b))
else:
    print("Invalid choice")


```
