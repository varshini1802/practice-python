```python
#Write a program to calculate BMI and print the category

height=float(input("enter your height in meters:"))
weight=float(input("enter your weight in kgs:"))
bmi=weight/(height**2)
print("your bmi is ",round(bmi,2))

if bmi<18.5:
    print("Underweight")
elif 18.5<=bmi<25:
    print("normal(healthy)")
elif 25<=bmi<30:
    print("overweight")
else:
    print("obesity")
```
