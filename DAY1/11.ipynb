```python
a = float(input("Enter first number: "))
op = input("Enter operator (+, -, *, /): ")
b = float(input("Enter second number: "))

if (op=="+"):
    print("result=", a+b)
if (op=="-"):
    print("result=", a-b)
if (op=="*"):
    print("result=", a*b)
if (op=="/"):
    if b==0:
        print("invalid")
    else:
        print("result=", a/b)   

```
