```python
#Build a 'Guess the Number' game with a fixed number and limited attempts using a loop

num=9

for i in range(5):
    guess=int(input("guess the number"))

    if guess==num:
        print("correct!!")
        break
    elif guess<num:
        print("Guess higher")
    else:
        print("guess lower")
else:
    print("you are out of attempts. The number was", num)
```
